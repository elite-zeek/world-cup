```python
from sklearn.model_selection import train_test_split\

x_train,x_test,y1_train,y1_test = train_test_split(x,y1,test_size = 0.2,random_state = 43)
x_train, x_test, y2_train, y2_test = train_test_split(x, y2, test_size=0.2, random_state=42)
```

# TRAIN OUR FIRST MODEL(REGRESSION)

```python
from sklearn.linear_model import LinearRegression

model1 = LinearRegression()

model1.fit(x_train,y1_train)
```

# TRAIN THE REGRESSION MODEL

```python
from sklearn.ensemble import RandomForestClassifier 

model2 = RandomForestClassifier(random_state = 42)

model2.fit(x_train,y2_train)
```

```python
y1_pred = model1.predict(x_test)
y2_pred = model2.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error,r2_score

print('MAE:',mean_absolute_error(y1_test,y1_pred))
print('MSE:',mean_squared_error(y1_test,y1_pred))
print('RMSE:',root_mean_squared_error(y1_test,y1_pred))
print('R2_score:',r2_score(y1_test,y1_pred))

```
